```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)
```

```python
x = np.linspace(-10, 10, 500)

sigmoid_y = sigmoid(x)
tanh_y = tanh(x)
relu_y = relu(x)

plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(x, sigmoid_y)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.subplot(3, 1, 2)
plt.plot(x, tanh_y)
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(x, relu_y)
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.plot(x, sigmoid_derivative(x))
plt.title("Sigmoid Gradient")
plt.grid()

plt.subplot(3, 1, 2)
plt.plot(x, tanh_derivative(x))
plt.title("Tanh Gradient")
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(x, relu_derivative(x))
plt.title("ReLU Gradient")
plt.grid()

plt.tight_layout()
plt.show()
```

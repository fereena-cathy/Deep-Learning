```python
#TASK A

import nltk
import re
import string
import numpy as np

from nltk.corpus import brown, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer


nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

print("Libraries imported and NLTK resources downloaded successfully.")
```

```python

# Load sentences from the Brown Corpus
sentences = brown.sents(categories='news')

print("Total sentences:", len(sentences))
print("\nSample original sentences:")

for sentence in sentences[:5]:
    print(" ".join(sentence))
```

```python

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

processed_sentences = []

for sentence in sentences:
    # Convert words to lowercase
    text = " ".join(sentence).lower()

    # Tokenize text
    tokens = word_tokenize(text)

    # Remove punctuation, non-alphabetic tokens, and stop words
    tokens = [
        word for word in tokens
        if word.isalpha()
        and word not in stop_words
    ]

    # Lemmatize words
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # Keep non-empty sentences
    if tokens:
        processed_sentences.append(tokens)

print("Original sentence:")
print(" ".join(sentences[0]))

print("\nPreprocessed sentence:")
print(" ".join(processed_sentences[0]))
```

```python
#TASK B
student_name = "Fereena"
roll_number ="24BADO24"

print(f"Student Name: {student_name}")
print(f"Roll Number: {roll_number}")

from nltk.probability import FreqDist


try:
    processed_sentences
except NameError:
    print("Warning: 'processed_sentences' was not detected in active memory. Re-generating tokens...")
    import nltk

    # Dynamically download resources if missing
    for resource in ['brown', 'punkt', 'punkt_tab', 'stopwords', 'wordnet', 'omw-1.4']:
        try:
            nltk.data.find(f"corpora/{resource}" if resource in ['brown', 'stopwords', 'wordnet'] else f"tokenizers/{resource}")
        except LookupError:
            nltk.download(resource, quiet=True)

    from nltk.corpus import brown
    from nltk.tokenize import word_tokenize
    from nltk.stem import WordNetLemmatizer
    from nltk.corpus import stopwords

    sentences = brown.sents(categories='news')
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()
    processed_sentences = []

    for sentence in sentences:
        text = " ".join(sentence).lower()
        tokens = word_tokenize(text)
        tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
        tokens = [lemmatizer.lemmatize(word) for word in tokens]
        if tokens:
            processed_sentences.append(tokens)

# Combine all preprocessed sentences into one list
all_words = [
    word
    for sentence in processed_sentences
    for word in sentence
]

# Create vocabulary
vocabulary = set(all_words)

# Calculate word frequencies
frequency_distribution = FreqDist(all_words)

print("Total number of words:", len(all_words))
print("Vocabulary size:", len(vocabulary))

print("\nTop 10 most frequent words:")
for word, frequency in frequency_distribution.most_common(10):
    print(word, ":", frequency)
```

```python

# Word2Vec requires a list of tokenized sentences
training_sentences = processed_sentences

print("Number of training sentences:", len(training_sentences))
print("\nSample tokenized sentences:")

for sentence in training_sentences[:5]:
    print(sentence)
```

```python
#task C

from gensim.models import Word2Vec

# Train Word2Vec using Skip-gram
model = Word2Vec(
    sentences=training_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=2,
    epochs=20,
    seed=42
)

print("Word2Vec model trained successfully!")
print("Vocabulary size:", len(model.wv))
print("Embedding dimension:", model.vector_size)
```

```python

# Select a word from the trained vocabulary
selected_word = "government"

if selected_word in model.wv:
    vector = model.wv[selected_word]

    print("Selected word:", selected_word)
    print("\nEmbedding vector:")
    print(vector)

    print("\nVector dimensions:", vector.shape)
    print("Number of values:", len(vector))
else:
    print("Word not found. Choose another word from the vocabulary.")
```

```python

if selected_word in model.wv:
    similar_words = model.wv.most_similar(
        selected_word,
        topn=5
    )

    print("Five words most similar to:", selected_word)

    for word, score in similar_words:
        print(f"{word}: {score:.4f}")
```

```python
#task D

# Candidate word pairs
candidate_pairs = [
    ("government", "state"),
    ("president", "government"),
    ("city", "town"),
    ("country", "state"),
    ("man", "woman"),
    ("year", "day"),
    ("house", "home")
]

# Select the first three pairs whose words are in the vocabulary
valid_pairs = [
    (word1, word2)
    for word1, word2 in candidate_pairs
    if word1 in model.wv and word2 in model.wv
][:3]

if len(valid_pairs) < 3:
    print("Fewer than three candidate pairs were found.")
    print("Choose additional words from the trained vocabulary.")
else:
    results = []

    for word1, word2 in valid_pairs:
        score = model.wv.similarity(word1, word2)
        results.append((word1, word2, score))

    print("Cosine Similarity Results")
    print("-" * 45)

    for word1, word2, score in results:
        print(f"{word1} - {word2}: {score:.4f}")

    most_similar_pair = max(results, key=lambda item: item[2])

    print("\nMost similar pair:")
    print(
        f"{most_similar_pair[0]} - {most_similar_pair[1]} "
        f"(Score: {most_similar_pair[2]:.4f})"
    )
```

```python

import pandas as pd

results_df = pd.DataFrame(
    results,
    columns=["Word 1", "Word 2", "Cosine Similarity"]
)

results_df = results_df.sort_values(
    by="Cosine Similarity",
    ascending=False
)

display(results_df)

results_df.to_csv("word_similarity_results.csv", index=False)

print("Results saved to word_similarity_results.csv")
```
